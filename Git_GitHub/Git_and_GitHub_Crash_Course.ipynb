## Staging Files with Git add

Create a folder *basics* and a new file called *example.txt*. Write *example* inside this file.

Let's check with file have changed inside *git-cloud-examples* using *git status*.
<figure>
    <img src="images/Git_and_Github_11.png" alt="Untracked Changes" title="Untracked Changes" width="5000" height="5000">
    <figcaption>Fig 11. Untracked Changes</figcaption>
</figure>
<br>
</br>

The old files came from the remote repository we cloned earlier, so Git already knows about them. But since ***basics*** is newly created, Git doesn't recognize it yet.

If we want to mode everything to the staging area at one, we can use: *git add --all* or *git add -A*.
<figure>
    <img src="images/Git_and_Github_12.png" alt="Stage Changes" title="Stage Changes" width="5000" height="5000">
    <figcaption>Fig 12. Stage Changes</figcaption>
</figure>
<br>
</br>

You'll see that Git has staged everything i.e., all the chasnges we made are now ready to be commited.

Now suppose we want to go back to the previous state -that is, remove everything from staging area and return the changes to the working directory, run: *git reset*

As soon as we run this command, everything goes back to the way it was before.

In the terminal, you'll see amessage saying *"Unstaged changes after reset"*. If we check the status again, everything is back to the earlier state like before.
<figure>
    <img src="images/Git_and_Github_13.png" alt="Reset Changes" title="Reset Changes" width="5000" height="5000">
    <figcaption>Fig 13. Reset Changes</figcaption>
</figure>
<br>
</br>

Now let's run: *git add -A*. Check the status. You'll set that Git has staged everything again like before. That means all the changes we made are ready to be commited.
<figure>
    <img src="images/Git_and_Github_14.png" alt="Staged Changes" title="Staged Changes" width="5000" height="5000">
    <figcaption>Fig 14. Staged Changes</figcaption>
</figure>
<br>
</br>

Now, run *git reset* again to unstage everything and bring all the changes back to the working directory. Then check the status. Everything should be back to the unstaged state.
<figure>
    <img src="images/Git_and_Github_15.png" alt="Reset Again" title="Reset Again" width="5000" height="5000">
    <figcaption>Fig 15. Reset Again</figcaption>
</figure>
<br>
</br>

Now, in this tate, run: *git add.*. At first, it may seem like it does the same thing as *git add -all* because everithing appears to be staged.
<figure>
    <img src="images/Git_and_Github_16.png" alt="Git add ." title="Git add ." width="5000" height="5000">
    <figcaption>Fig 16. Git add .</figcaption>
</figure>
<br>
</br>

But that is not always the case. Let's run *git reset* again to return everything to the previous state, and we'll see an importnat difference. After *git reset*, all files are back in the working directory. Next, go inside the **basics** directory and run: *git add .*. Run "git status* and notice that Git has only staged *example.txt* inside **basics**.

The other files in the root folder are still unstaged.
<figure>
    <img src="images/Git_and_Github_17.png" alt="Git add . Inside Basics" title="Git add . Inside Basics" width="5000" height="5000">
    <figcaption>Fig 17. Git add . Inside Basics</figcaption>
</figure>
<br>
</br>

Also, keep in mind that if there were a subfolder inside **basics**, using the dot would still include the files inside that subfolder as well. In simple terms, the dot represents the current directory **and everything inside it**.

So now we've seen three variations of the *git add* command: </br>
*git add --all* </br>
*git add -A* </br>
*git add .* </br>

Now, run the *git reset* to bring everything back to the previous state again. And go back to the root folder: *cd ..*.

Then stage all files with: *git add --all*. Now, all our files are staged and ready to be commited.

Manually delete the *notes.txt* file and create a new one named *minutes.txt*, writing *minutes* inside it.
<figure>
    <img src="images/Git_and_Github_18.png" alt="Commited and Not Staged for Commit" title="Commited and Not Staged for Commit" width="5000" height="5000">
    <figcaption>Fig 18. Commited and Not Staged for Commit</figcaption>
</figure>
<br>
</br>

I've deleted one file and added a new one. Git shows that under *Changes to be commited*, the previously staged files are still listed. But under *Changes not staged for commit*, it now lists *notes.txt* as deleted and *minutes.txt* as untracked.

Now run: *git add \**. Then check the status. You'll notice that Git has staged the newly created *minutes.txt* file, but it has **not** staged the deleted *notes.txt* file.
<figure>
    <img src="images/Git_and_Github_19.png" alt="git add *" title="git add *" width="5000" height="5000">
    <figcaption>Fig 19. git add *</figcaption>
</figure>
<br>
</br>

This is an important detail, *git add \** stages **new and modified** files, but it does **not** stage deleted giles. In other words, deleted files are ignored by this command.

That'sd the key difference between using *\** and using *.*, or *-A* with *git add*. The star (*) stages all visible files, but it does not include deletions.

Now, let's return to the previous state again by typing: *git reset*.

Now, if you want to stage only a specific file --for example, *hello.txt*, which has been modified-- you can do that by typing: *git add hello.txt*.
<figure>
    <img src="images/Git_and_Github_20.png" alt="Stage an Specific File" title="Stage an Specific File" width="5000" height="5000">
    <figcaption>Fig 20. Stage an Specific File</figcaption>
</figure>
<br>
</br>

You can also stage a file inside a folder, such as inside **basics** for example: *git add bsics/example.txt*

Let's try staging our delete of *notes.txt* by running: *git add notes.txt*

If we check that status using, we'll see that only *notes.txt* has been staged.
<figure>
    <img src="images/Git_and_Github_21.png" alt="Stage a Delete File" title="Stage a Delete File" width="5000" height="5000">
    <figcaption>Fig 21. Stage a Delete File</figcaption>
</figure>
<br>
</br>

That's how we stage individual files by specifying their names.

Now let's return everything to the previous **unstaged** state by running: *git reset*

We can also stage files by their extension. For example, to stage all *.txt* file, we write: *git add \*.txt*

This tages all *.txt* files in the **current directory**. Note that it won't include deleted files or files inside subfolders.

<figure>
    <img src="images/Git_and_Github_22.png" alt="Stage Files by Extension" title="Stage Files by Extension" width="5000" height="5000">
    <figcaption>Fig 22. Stage Files by Extension</figcaption>
</figure>
<br>
</br>

If we want to stage everything at once, the best and simplest practice is to go to the **root directory** and type: *git add .*

After doing that, all changes are staged together.

If we now check the status, we'll see everything has been staged successfully. That means all our changes have been moved from the **working directory** to the **staging area**.

<figure>
    <img src="images/Git_and_Github_23.png" alt="Stage All" title="Stage All" width="5000" height="5000">
    <figcaption>Fig 22. Stage All</figcaption>
</figure>
